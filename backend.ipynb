```python
from docx import Document
import os

def extract_docx_text(filepath):
    doc = Document(filepath)

    text = []

    # Paragraphs
    for paragraph in doc.paragraphs:
        if paragraph.text.strip():
            text.append(paragraph.text.strip())

    # Tables
    for table in doc.tables:
        for row in table.rows:
            row_text = []

            for cell in row.cells:
                if cell.text.strip():
                    row_text.append(cell.text.strip())

            if row_text:
                text.append(" | ".join(row_text))

    return "\n".join(text)


resume_data = []

for filename in os.listdir(resume_folder):

    if filename.lower().endswith(".docx"):

        filepath = os.path.join(resume_folder, filename)

        try:
            text = extract_docx_text(filepath)

            resume_data.append({
                "filename": filename,
                "text": text
            })

        except Exception as e:
            print(f"Error reading {filename}: {e}")

print("Total resumes extracted:", len(resume_data))
```

```python
import re

def clean_text(text):

    text = text.lower()

    # Replace multiple spaces
    text = re.sub(r"\s+", " ", text)

    # Keep letters, numbers and common symbols
    text = re.sub(r"[^a-z0-9+#.\- ]", " ", text)

    # Remove repeated spaces
    text = re.sub(r"\s+", " ", text)

    return text.strip()


df["clean_text"] = df["text"].apply(clean_text)

df[["filename", "clean_text"]].head()
```

```python
job_description=job_description = """
We are looking for a Machine Learning Engineer.

Requirements:
- Bachelor's or Master's degree in Computer Science, Artificial Intelligence,
  Data Science or related field.
- Minimum 3 years of experience in Machine Learning.
- Strong Python programming skills.
- Experience with PyTorch and TensorFlow.
- Strong SQL knowledge.
- Experience with NLP and Deep Learning.
- Experience with Docker is preferred.
"""
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [job_description] + df["clean_text"].tolist()

tfidf = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2),
    max_features=10000
)

tfidf_matrix = tfidf.fit_transform(documents)

job_vector = tfidf_matrix[0]

resume_vectors = tfidf_matrix[1:]

tfidf_scores = cosine_similarity(
    job_vector,
    resume_vectors
).flatten()

df["tfidf_score"] = tfidf_scores

df[["filename", "tfidf_score"]].sort_values(
    "tfidf_score",
    ascending=False
).head(10)
```

open source embedding model

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)
```

```python
job_embedding = embedding_model.encode(
    job_description,
    normalize_embeddings=True
)

print(job_embedding.shape)
```

```python
resume_embeddings = embedding_model.encode(
    df["text"].tolist(),
    normalize_embeddings=True,
    show_progress_bar=True
)

print(resume_embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

semantic_scores = cosine_similarity(
    job_embedding.reshape(1, -1),
    resume_embeddings
).flatten()

df["semantic_score"] = semantic_scores
```
